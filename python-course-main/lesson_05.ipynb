### Part 1: What is a Loop?
A loop repeats code multiple times. The while loop continues as long as a condition is True.  
Syntax:

```python
while condition:
    # code to repeat
    # this code runs while condition is True
```

Code cell:


```python
# Simple while loop
count = 1

while count <= 5:
    print(f"Count is: {count}")
    count = count + 1

print("Loop finished!")
```

--- 
### Part 2: Loop with User Input
Code cell:

```python
# Keep asking until user types "stop"
while True:
    answer = input("Type 'stop' to exit: ")
    
    if answer == "stop":
        break
    
    print("You typed:", answer)

print("Program ended")
```

---
### Part 3: The break Statement
The break statement exits the loop immediately.  
Code cell:

```python
# Find first number divisible by 7
number = 1

while number <= 100:
    if number % 7 == 0:
        print(f"First number divisible by 7 is: {number}")
        break
    number = number + 1
```

---
### Part 4: The continue Statement
The continue statement skips to the next iteration.   
Code cell:

```python
# Print only odd numbers from 1 to 10
number = 0

while number < 10:
    number = number + 1
    
    if number % 2 == 0:
        continue  # Skip even numbers
    
    print(number)
```

---
### Part 5: Common Patterns
Pattern 1: Counter

```python
count = 0
while count < 5:
    print("Hello")
    count = count + 1
```

Pattern 2: User Input Validation


```python
age = -1
while age < 0:
    age = int(input("Enter your age: "))
    if age < 0:
        print("Age cannot be negative!")
```

Pattern 3: Menu System


```python
choice = ""
while choice != "quit":
    print("1. Option A")
    print("2. Option B")
    print("Type 'quit' to exit")
    choice = input("Choose: ")
```

--- 
### Important Notes

- The condition is checked before each iteration
- Make sure the condition will eventually become False
- Always update the variable in the condition to avoid infinite loops
- Use break to exit early
- Use continue to skip to next iteration

### Task 2: Sum Calculator
Write a program that:

- Asks the user to enter numbers
- Keeps adding them to a total
- Stops when user enters 0
- Prints the final sum

Code cell:

```python
# Your code here
total = 0
# Start an infinite loop that we will break out of manually
while True:
    try:
        # Ask the user for input and convert it to an integer
        number_input = int(input("Enter a number (enter 0 to stop): "))
        
        # Check if the user entered 0 to stop the loop
        if number_input == 0:
            break  # Exit the while loop
        
        # Add the number to the running total
        total += number_input
        print(f"Current total is: {total}") # Optional: show running total
        
    except ValueError:
        # Handle cases where the user types text instead of a number
        print("Invalid input. Please enter a valid integer.")

print(f"The final sum is: {total}")
```

### Task 4: Guessing Game
Write a number guessing game:

- Computer picks a number between 1 and 10 (you choose it)
- User keeps guessing
- Give hints: "Too high" or "Too low"
- When correct, print how many guesses it took

Code cell:

```python
# Your code here
secret_number = 7
guesses = 0
while guesses < 5:
    try:
        user_guess = int(input("Guess the secret number (between 1 and 10): "))
        
        if user_guess == secret_number:
            print("Congratulations! You've guessed the correct number.")
            break
        else:
            print("Wrong guess. Try again.")
            guesses += 1
    except ValueError:
        print("Please enter a valid integer.")
```

### Task 5: Menu System
Create a simple calculator menu:

- Show options: 1-Add, 2-Subtract, 3-Multiply, 4-Divide, 5-Exit
- Ask user to choose
- If 1-4: ask for two numbers and show result
- If 5: exit the program
- Keep showing menu until user exits

Code cell:

```python
# Your code here
while True:
    print("\n--- Calculator Menu ---")
    print("1-Add")
    print("2-Subtract")
    print("3-Multiply")
    print("4-Divide")
    print("5-Exit")
    print("-----------------------")
    
    # Get user choice
    choice = input("Enter your choice (1-5): ")
    
    # Check if choice is 5 to exit
    if choice == '5':
        print("Exiting program. Goodbye!")
        break
    
    # Check if choice is a valid operation (1-4)
    if choice in ['1', '2', '3', '4']:
        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
            
            if choice == '1':
                result = num1 + num2
                print(f"Result: {num1} + {num2} = {result}")
            elif choice == '2':
                result = num1 - num2
                print(f"Result: {num1} - {num2} = {result}")
            elif choice == '3':
                result = num1 * num2
                print(f"Result: {num1} * {num2} = {result}")
            elif choice == '4':
                # Handle division by zero
                if num2 == 0:
                    print("Error: Cannot divide by zero.")
                else:
                    result = num1 / num2
                    print(f"Result: {num1} / {num2} = {result}")
                    
        except ValueError:
            print("Invalid input. Please enter valid numbers.")
    else:
        print("Invalid choice. Please enter a number between 1 and 5.")
```

### Task 6: Even Numbers Only (Challenge)
Write a program that:

- Asks the user to enter numbers
- Only accepts even numbers
- If odd, print "Only even numbers!" and ask again
- Stops when user enters 0
- Prints all the even numbers entered

Code cell:

```python
# Your code here
even_numbers = []
print("Enter numbers (0 to stop). Only even numbers will be accepted.")

while True:
    try:
        # Get input and convert to integer
        user_input = int(input("Enter a number: "))
        
        # Check for exit condition (0)
        if user_input == 0:
            break
        
        # Check if the number is even using the modulo operator (%)
        # If the remainder when divided by 2 is 0, it's even.
        if user_input % 2 == 0:
            even_numbers.append(user_input)
            print(f"Accepted: {user_input}")
        else:
            print("Only even numbers!")
            
    except ValueError:
        print("Invalid input. Please enter an integer.")

print(f"\nAll even numbers entered: {even_numbers}")
```
